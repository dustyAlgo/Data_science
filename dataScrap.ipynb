```python
import requests
from bs4 import BeautifulSoup

base_url = "https://books.toscrape.com/catalogue/"
page_url = "https://books.toscrape.com/catalogue/page-1.html"

while True:
    # 1️⃣ Fetch the page
    response = requests.get(page_url)
    soup = BeautifulSoup(response.text, "html.parser")

    # 2️⃣ Extract all book titles on this page
    books = soup.select("ol.row article.product_pod a")
    for book in books:
        print(book["href"])

    # 3️⃣ Find the next page link (if any)
    next_page = soup.select_one("li.next a")
    if next_page:
        next_href = next_page["href"]
        # Build full URL for the next page
        page_url = base_url + next_href
    else:
        # No next page → stop loop
        break
```
